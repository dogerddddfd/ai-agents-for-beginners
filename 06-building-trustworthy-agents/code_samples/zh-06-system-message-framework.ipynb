```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential
from dotenv import load_dotenv

# 从 .env 文件加载环境变量
load_dotenv()

# 从环境变量中获取模型、端点和密钥
endpoint = os.getenv("API_URL")
token = os.getenv("API_KEY")
```

```python
# 指定使用的模型名称
model_name = os.getenv("MODEL_FREE_8B")

# 创建ChatCompletionsClient客户端
client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)
```

```python
# 定义角色、公司和职责
role = "travel agent"  # 旅游代理
company = "contoso travel"  # 康托索旅游公司
responsibility = "booking flights"  # 预订航班
```

```python
# 使用Azure AI生成系统提示
response = client.complete(
    messages=[
        SystemMessage(content="""你是创建AI代理助手的专家。
你将获得公司名称、角色、职责和其他信息，用于创建系统提示。
创建系统提示时，请尽可能详细，并提供一个结构，使使用LLM的系统能够更好地理解AI助手的角色和职责。"""),
        UserMessage(content=f"你是{role}，在{company}工作，负责{responsibility}。"),
    ],
    model=model_name,
    # 可选参数
    temperature=1.,
    max_tokens=1000,
    top_p=1.
)

# 打印生成的系统提示
print(response.choices[0].message.content)
```
