# 🐍 使用 Microsoft Agent Framework (Python) 的 AI 旅行代理

## 📋 场景概述

本笔记本演示了如何使用 Microsoft Agent Framework for Python 构建智能旅行规划代理。该代理利用 GitHub Models 自动为全球随机目的地生成个性化的一日游行程。

**主要功能：**
- 🎲 **智能目的地选择**：用于随机目的地选择的自定义工具函数
- 🗺️ **详细行程生成**：带有本地推荐的 AI 驱动的旅行规划
- 🔄 **异步处理**：使用 asyncio 进行高效的 API 通信
- 🛠️ **工具集成**：展示 AI 代理中的函数调用能力

## 🏗️ 技术实现

### 核心组件
- **Agent Framework**：Microsoft 代理编排系统的 Python 实现
- **GitHub Models API**：通过 GitHub 的推理服务访问最先进的语言模型
- **OpenAI 兼容性**：使用带有 GitHub Models 后端的 OpenAI 客户端模式
- **环境管理**：使用 python-dotenv 安全处理凭证

### 架构流程
```python
用户请求 → ChatAgent → GitHub Models API ↔ get_random_destination()
                     ↓
              旅行行程响应
```

### 关键类和方法
- `ChatAgent`：主要的会话代理编排器
- `OpenAIChatClient`：GitHub Models API 客户端包装器
- `get_random_destination()`：用于目的地选择的自定义工具函数
- 环境变量：安全的 API 配置管理

````python
# 📦 导入所需库
# 用于系统操作和随机数生成的标准库导入
import os
from random import randint

# 用于从 .env 文件加载环境变量的第三方库
from dotenv import load_dotenv
````

````python
# 🤖 导入 Microsoft Agent Framework 组件
# ChatAgent：会话 AI 的主要代理类
# OpenAIChatClient：用于连接 OpenAI 兼容 API 的客户端（包括 GitHub Models）
from agent_framework import ChatAgent
from agent_framework.openai import OpenAIChatClient
````

````python
# 🔧 加载环境变量
# 这会从项目根目录的 .env 文件加载配置
# 所需变量：GITHUB_ENDPOINT, GITHUB_TOKEN, GITHUB_MODEL_ID
load_dotenv()
````

````python
# 🎲 工具函数：随机目的地生成器
# 此函数将作为工具提供给代理
# 代理可以调用此函数获取随机度假目的地
def get_random_destination() -> str:
    """获取随机度假目的地。
    
    返回：
        str: 从我们预定义列表中随机选择的目的地
    """
    # 世界各地热门度假目的地列表
    print("是的孩子们get_random_destination被调用了")
    destinations = [
        "Barcelona, Spain",
        "Paris, France", 
        "Berlin, Germany",
        "Tokyo, Japan",
        "Sydney, Australia",
        "New York, USA",
        "Cairo, Egypt",
        "Cape Town, South Africa",
        "Rio de Janeiro, Brazil",
        "Bali, Indonesia"
    ]
    # 从列表中返回随机目的地
    return destinations[randint(0, len(destinations) - 1)]
````

````python
# 🔗 为 GitHub Models 创建 OpenAI 聊天客户端
# 此客户端连接到 GitHub Models API（OpenAI 兼容端点）
# 所需环境变量：
# - GITHUB_ENDPOINT: API 端点 URL（通常为 https://models.inference.ai.azure.com）
# - GITHUB_TOKEN: 您的 GitHub 个人访问令牌
# - GITHUB_MODEL_ID: 要使用的模型（例如，gpt-4o-mini, gpt-4o）
# openai_chat_client = OpenAIChatClient(
#     base_url=os.environ.get("GITHUB_ENDPOINT"),
#     api_key=os.environ.get("GITHUB_TOKEN"), 
#     model_id=os.environ.get("GITHUB_MODEL_ID")
# )
````

````python
# 使用其他服务商
openai_chat_client = OpenAIChatClient(
    base_url=os.environ.get("API_URL"),
    api_key=os.environ.get("API_KEY"), 
    model_id=os.environ.get("MODEL_FREE_8B")
    # model_id=os.environ.get("MODEL_32B")
)
````

````python
# 🤖 创建旅行规划代理
# 这将创建具有特定功能的会话 AI 代理：
# - chat_client: 用于生成响应的 AI 模型客户端
# - instructions: 定义代理个性和角色的系统提示
# - tools: 代理可以调用以执行操作的函数列表
agent = ChatAgent(
    chat_client=openai_chat_client,
    instructions="You are a helpful AI Agent that can help plan vacations for customers at random destinations.",
    tools=[get_random_destination]  # 我们的随机目的地工具函数
)
````

````python
# 🚀 运行代理
# 向代理发送消息并获取响应
# 代理将在需要时使用其工具（get_random_destination）
response = await agent.run("Plan me a day trip")
````

````python
# 📋 查看原始响应对象
# 这显示完整的响应结构，包括元数据
# 对调试和理解响应格式很有用
print(response)
````

````python
# 📖 提取并显示旅行计划
# 获取对话中的最后一条消息（代理的响应）
last_message = response.messages[-1]
# 从消息中提取文本内容
text_content = last_message.contents[0].text
# 显示格式化的旅行计划
print("🏖️ 旅行计划：")
print(text_content)
````
