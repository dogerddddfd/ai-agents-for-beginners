````python
# 📚 探索智能体框架 - 基础智能体示例
# 本示例展示了Microsoft智能体框架的核心概念

# 📦 导入所需库
import os                     # 用于访问环境变量
from random import randint    # 用于生成随机选择

from dotenv import load_dotenv  # 用于加载.env配置文件
````

````python
# 🤖 导入核心智能体框架组件
# ChatAgent: 主要的对话智能体类
# OpenAIChatClient: 用于连接兼容OpenAI的API的客户端
from agent_framework import ChatAgent
from agent_framework.openai import OpenAIChatClient
````

````python
# 🔧 初始化环境配置
# 从.env文件加载环境变量
# 这实现了API密钥和端点的安全存储
load_dotenv()
````

````python
# 🛠️ 工具函数：随机目的地生成器
# 这展示了如何创建智能体可以使用的工具
# 工具扩展了智能体的能力，超越了单纯的对话
def get_random_destination() -> str:
    """获取随机度假目的地。
    
    此函数作为智能体可以调用的工具，当它需要
    为旅行规划建议随机目的地时使用。
    
    返回：
        str: 从预定义列表中随机选择的目的地
    """
    # 全球热门度假目的地精选列表
    destinations = [
        "Barcelona, Spain",
        "Paris, France", 
        "Berlin, Germany",
        "Tokyo, Japan",
        "Sydney, Australia",
        "New York, USA",
        "Cairo, Egypt",
        "Cape Town, South Africa",
        "Rio de Janeiro, Brazil",
        "Bali, Indonesia"
    ]
    # 返回随机选择的目的地
    return destinations[randint(0, len(destinations) - 1)]
````

````python
# 🔗 创建OpenAI聊天客户端
# 这展示了如何连接到GitHub模型（兼容OpenAI的API）
# 客户端处理与AI模型的所有通信
# openai_chat_client = OpenAIChatClient(
#     base_url=os.environ.get("GITHUB_ENDPOINT"),    # API端点URL
#     api_key=os.environ.get("GITHUB_TOKEN"),        # 认证令牌
#     model_id=os.environ.get("GITHUB_MODEL_ID")  # 模型标识符（例如，gpt-4o-mini）
# )

openai_chat_client = OpenAIChatClient(
    base_url=os.environ.get("API_URL"),
    api_key=os.environ.get("API_KEY"), 
    model_id=os.environ.get("MODEL_FREE_8B")
    # model_id=os.environ.get("MODEL_32B")
)
````

````python
# 🤖 创建基础智能体
# 这展示了框架中的核心智能体创建模式
# 关键组件：聊天客户端、指令和工具
agent = ChatAgent(
    chat_client=openai_chat_client,  # 用于生成响应的AI模型客户端
    instructions="你是一个有帮助的AI智能体，可以帮助客户在随机目的地规划度假。",  # 定义智能体行为的系统提示
    tools=[get_random_destination]   # 智能体可以使用的可用工具列表
)
````

````python
thread = agent.get_new_thread()
````

````python
response1 = await agent.run("为我规划一次日间旅行",thread= thread)
````

````python

last_message = response1.messages[-1]
text_content = last_message.contents[0].text
print("旅行计划:")
print(text_content)
````

````python
response2 = await agent.run("我不喜欢那个目的地。为我规划另一个度假。",thread= thread)
````

````python
last_message = response2.messages[-1]
text_content = last_message.contents[0].text
print("变更计划:")
print(text_content)
````
