````python
# 📦 导入标准库
# os: 用于环境变量访问
# dotenv: 用于从 .env 文件加载环境变量
import os

from dotenv import load_dotenv
````

````python
# 📝 导入数据建模库
# Pydantic: 用于数据验证和结构化输出
# BaseModel: 用于创建数据模型的基类
# Field: 用于向模型字段添加元数据和验证
# List: 列表集合的类型提示
from pydantic import BaseModel, Field
from typing import List
````

````python
# 🤖 导入 Microsoft Agent Framework 组件
# ChatMessage: 用于创建结构化聊天消息
# Role: 消息角色的枚举（用户、助手、系统）
# ChatOptions: 聊天交互的配置选项
# OpenAIChatClient: OpenAI 兼容 API 的客户端（GitHub Models）
from agent_framework import ChatMessage, Role, ChatOptions
from agent_framework.openai import OpenAIChatClient
````

````python
# 🔧 加载环境变量
# 从 .env 文件加载配置
# 必需：GITHUB_ENDPOINT, GITHUB_TOKEN, GITHUB_MODEL_ID
load_dotenv()
````

````python
# 🎯 代理配置
# 定义代理的名称和角色

AGENT_NAME = "TravelAgent"

# 📋 规划代理的系统指令
# 此代理充当协调者，决定使用哪些专业代理
AGENT_INSTRUCTIONS = """You are a planner agent.
    Your job is to decide which agents to run based on the user's request.
    Below are the available agents specialized in different tasks:
    - FlightBooking: For booking flights and providing flight information
    - HotelBooking: For booking hotels and providing hotel information
    - CarRental: For booking cars and providing car rental information
    - ActivitiesBooking: For booking activities and providing activity information
"""
````

````python
class SubTask(BaseModel):
    assigned_agent: str = Field(
        description="The specific agent assigned to handle this subtask")
    task_details: str = Field(
        description="Detailed description of what needs to be done for this subtask")


class TravelPlan(BaseModel):
    main_task: str = Field(
        description="The overall travel request from the user")
    subtasks: List[SubTask] = Field(
        description="List of subtasks broken down from the main task, each assigned to a specialized agent")
````

````python
options = ChatOptions(response_format=TravelPlan)
````

````python
client = OpenAIChatClient(base_url=os.environ.get("GITHUB_ENDPOINT"), api_key=os.environ.get("GITHUB_TOKEN"), model_id=os.environ.get("GITHUB_MODEL_ID"))
````

````python
agent =  client.create_agent(name= AGENT_NAME , instructions=AGENT_INSTRUCTIONS)
````

````python
messages = [
        ChatMessage(role=Role.USER, text="Create a travel plan for a family of 4, with 2 kids, from Singapore to Melbourne")
    ]
````

````python
response = await agent.run(messages,response_format=TravelPlan)
````
